## Dataset 1
- **Source:** [PPE_DATASET_YOLOv8](https://www.kaggle.com/datasets/shlokraval/ppe-dataset-yolov8)
- **License:** Apache 2.0
- **Classes used:** Fall-Detected, Gloves, Goggles, Hardhat, Ladder, Mask, NO-Gloves, NO-Goggles, NO-Hardhat, NO-Mask, NO-Safety Vest, Person, Safety Cone, Safety Vest

## Dataset 2 
- **Source:** [Industrial Hazards Detection](https://www.kaggle.com/datasets/vigneshnachu/industrial-hazards-detection)
- **License:** Attribution 4.0 International (CC BY 4.0)
- **Classes used:** Fire, Smoke, Water Leak, and Chemical Leak

## How to Train the model

1. Install the dependencies using `pip install requirements.txt`
2. Download the datasets. Create a folder for Dataset 1 called `hazard` and another folder for Dataset 2 called `ppe`. Extract their zip files in their corresponding folders.
3. Move the folders to `data/raw` (create a folder if there isn't a `raw` folder)
4. Run `python scripts/remove_no_helmet.py`
5. Run `python scripts/train_datasets.py`. Feel free to edit the parameters in the `export_model.py`.
6. Run `python scripts/export_model.py` (run it twice because you have to input each model). The model after training is found in the `runs/detect/<model-name>/weights/best.pt`. Make sure to give it only either filenames `hazard` or `ppe`.
7. Optionally `python scripts/validation.py` to get the model performance. 

Note: You may use Google Colab to speed up the training, but it will not be covered here.
